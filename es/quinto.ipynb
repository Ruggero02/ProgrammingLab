```python
%pip install pandas matplotlib seaborn
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
df.head()
```

```python
righe, colonne = df.shape
print(f"Il dataset contiene {righe} righe e {colonne} colonne.")
print("\nInformazioni sulle colonne:")
for col in df.columns:
    print(f"\nColonna: {col}")
    print(f"  Valori nulli: {df[col].isnull().sum()}")
```

```python
moda_embarked = df['Embarked'].mode()[0]
df['Embarked'] = df['Embarked'].fillna(moda_embarked)
print(f"\nValori mancanti in 'Embarked' compilati con la moda: '{moda_embarked}'")
```

```python
numero_duplicati = df.duplicated().sum()
print(f"\nNumero di righe duplicate: {numero_duplicati}")
```
