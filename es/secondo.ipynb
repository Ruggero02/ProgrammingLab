```python
freq = np.array([68, 65, 77, 110, 160, 161, 162, 161, 160, 161, 162, 163, 164, 163, 162, 100, 90, 97, 72, 60, 70])

print("Frequenza Minima (Riposo):", freq.min())
print("Frequenza Massima (Esercizio):", freq.max())

esercizio_mask = freq > 120
print("Array Booleano Esercizio:", esercizio_mask)

percentuale = np.mean(esercizio_mask) * 100
print(f"Percentuale tempo esercizio: {percentuale:.2f}%")
```

```python
stipendi = np.array([50000, 105250, 55000, 89000])
print("Costo totale iniziale:", stipendi.sum())


stipendi = np.array([50000, 105250, 55000, 89000], dtype=float)


stipendi[stipendi == 105250] *= 1.15
print("Nuovo stipendio CEO:", stipendi[1])


stipendi[stipendi == 50000] *= 1.20


stipendi[(stipendi != 121037.5) & (stipendi != 60000)] *= 1.10

print("Vettore stipendi finale:", stipendi)
print("Costo totale finale:", stipendi.sum())
```
