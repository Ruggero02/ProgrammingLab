```python
doppio = []
for i in range(5):
    doppio.append(i*2)

list_doppio = [i*2 for i in range(5)]
print(doppio)
print(list_doppio)
```

```python
pari = []
for i in range(10):
    if i%2==0:
        pari.append(i)
list_pari = [i for i in range(10) if i%2==0]
print(pari)
print(list_pari)
```

```python
cibo = ["mela", "carne", "pasta"]
cibo_maiusc = []
for c in cibo:
    cibo_maiusc.append(c.upper())
    
list_cibo = [c.upper() for c in cibo]
print(cibo_maiusc)
print(list_cibo)
```
